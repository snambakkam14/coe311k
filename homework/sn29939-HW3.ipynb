```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
x_values = np.arange(0,10.05,0.1)
```

```python
n_one = []
n_two = []
n_three = []
n_four = []

for x_val in x_values:
    n_one.append(sin_taylor(x_val,1))  
for x_val in x_values:
    n_two.append(sin_taylor(x_val,2))
for x_val in x_values:
    n_three.append(sin_taylor(x_val,3)) 
for x_val in x_values:
    n_four.append(sin_taylor(x_val,4))  
```

```python
actual_value = np.sin(x_values)
```

```python
absolute_error_one = np.abs(actual_value - n_one)
absolute_error_two = np.abs(actual_value - n_two)
absolute_error_three = np.abs(actual_value - n_three)
absolute_error_four = np.abs(actual_value - n_four)

plt.figure()
plt.plot(x_values,absolute_error_one,'b',label = "1 term")
plt.plot(x_values,absolute_error_two, 'y', label = "2 terms")
plt.plot(x_values,absolute_error_three, 'g', label = "3 terms")
plt.plot(x_values,absolute_error_four, 'r', label = "4 terms")
plt.xlabel("x values")
plt.ylabel("abs error")
plt.title("Absolute Error vs x_value")
plt.legend()
plt.grid(True)
plt.show()
```

The nature of the Taylor series is that it becomes more accurate as more terms are added. However, at a certain point, the error is more for greater terms due to the exponential nature of the series. as the x values get larger, the exponential term of the series will get larger at a really quick rate. The approximation is most accurate with more terms and a smaller x value close to the center of the series. In this case, the series is centered around x=0. The error does not decrease uniformly for all values of x as the rate of change in error for one term grows very slowly after 3-4 terms while the rate of change for absolute error in the 4 termed data continues to grow at a significant rate of change throughout the 10 terms.
