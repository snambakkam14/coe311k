Part 1 of Challenge

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import random

X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

A = np.vstack([X, np.ones(len(X))]).T

m_1, c_1 = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m_1}, Intercept: {c_1}")
```

```python
plt.figure()
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.scatter(X,Y)

x_best_fit = np.linspace(min(X),max(X),200)
y_best_fit = m_1*x_best_fit + c_1
plt.plot(x_best_fit,y_best_fit)
```

```python
X_2 = np.random.rand(8)
Y_2 = np.random.rand(8)


A_2 = np.vstack([X_2,np.ones(len(X_2))]).T

print("x =",X_2)
print("\ny =",Y_2)



m_2,c_2 = np.linalg.lstsq(A_2,Y_2, rcond=None)[0]
print(f"\nSlope: {m_2}, Intercept: {c_2}")
```

Part 3 of Challenge

```python
# y = mx + c

m_actual = 10
c_actual = 14
X_3 = np.arange(0,1.01,0.01)
err = 2*(np.random.rand(101)-0.5)
Y_3 = m_actual * X_3 + c_actual + err

A_3 = np.vstack([X_3, np.ones(len(X_3))]).T

print("A: ", A_3)
print("\nY: ", Y_3)
```

```python
M_3,C_3 = np.linalg.lstsq(A_3, Y_3, rcond=None)[0]
print(f"Slope: {M_3}, Intercept: {C_3}")
```

```python
plt.figure()
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)


x_actual = np.arange(0,1.01,0.01)
y_actual = x_actual * m_actual + c_actual
plt.plot(x_actual,y_actual, color = 'red', label = 'True equation')

x_best_fit = x_actual
y_best_fit = M_3 * x_best_fit + C_3
plt.plot(x_best_fit,y_best_fit, color = 'green', label = 'Line of best fit')



plt.scatter(X_3,Y_3)


plt.legend()
```
