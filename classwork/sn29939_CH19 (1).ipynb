```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(t,y):
    return (-1000*y + 3000 - 2000*np.exp(-t))

def exact_sol(t):
    return (3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t))

```

```python
t_exact = np.linspace(0,30,10000)    
y_exact = exact_sol(t_exact)
```

```python
h_arr_2 = [.1, .01, .001, .00001, .0000001]

for h in h_arr_2:
    t_start, t_end = 0,30
    num_steps = int((t_end - t_start) / h)

    t_values = np.linspace(t_start, t_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)


    # Initial condition
    y_values[0] = 0  # y(0) = 0


    # Implement Euler's method

    for i in range(num_steps):
        
        t_values[i+1] = t_values[i] + num_step
        
        y_values[i+1] = y_values[i] + num_step * f(t_values[i], y_values[i])
    
    
    plt.plot(t_values, y_values, label='Euler Forward')
    plt.plot(t_exact,y_exact, label = 'Exact Solution')
    


 

plt.xlabel('t')
plt.ylabel('y')
plt.legend()
plt.title('Solution of DE ')
plt.grid(True)
        
    
    
```
