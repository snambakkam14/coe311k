```python
import numpy as np
import matplotlib.pyplot as plt
import time as tm

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
            
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
            
        x = x_new
        
    return x, max_iterations
```

```python
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()

    for k in range(max_iterations):
        x_new = x.copy()

        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k

        x = x_new

    return x, max_iterations
```

```python
A = np.array([[10, -1, 2, 0, 0],
              [-1, 11, -1, 3, 0],
              [2, -1, 10, -1, 0],
              [0, 3, -1, 8, -2],
              [0, 0, 0, -2, 9]], dtype = 'float')

b = np.array([14, 30, 26, 25, 37], dtype = 'float')

x0 = np.zeros_like(b, dtype = 'float')

tol = 1e-6

max_iterations = 100


#jacobi


start_time = tm.time()

solution, iterations = jacobi(A, b, x0, tol, max_iterations)

end_time = tm.time()


print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {end_time - start_time} seconds")



#gauss seidel



start_time = tm.time()

solution, iterations = gauss_seidel(A, b, x0, tol, max_iterations)

end_time = tm.time()


print(f"SOLUTION: {solution}")
print(f"ITERATIONS: {iterations}")
print(f"TIME: {end_time - start_time} seconds")

```

When running both methods, they converged quickly with the Gauss-Seidel method running quicker and on less iterations, proving to be a more efficient method. The Gauss-Seidel method had 10 iterations with a 0.0016939640045166016 second run time while the Jacobi method had 18 iterations with a 0.0034203529357910156 second run time. The convergencewith both methods happen because the system is diagonally dominant.
