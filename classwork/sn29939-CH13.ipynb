```python
import numpy as np
import matplotlib.pyplot as plt
import time as tm
import random
```

```python
def naive_matrix_multiplication(A, B): # going to be used to go against A@B and compare
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
            
                C[i][j] += A[i][k] * B[k][j]
    return C
```

```python
a = 3

naive_runtime = np.zeros(8)
py_runtime = np.zeros(8)
matrix_size = np.zeros(8)

for i in range(8):
    
    size = a*(2**i)
    
    A = np.random.rand(size,size)
    B = np.random.rand(size,size)
    
    matrix_size[i] = size
    
    start = tm.time()
    C_naive = naive_matrix_multiplication(A, B)
    end = tm.time()
    runtime_naive = end - start
    naive_runtime[i] = runtime_naive
    
    
    start2 = tm.time()
    C_py = np.dot(A,B)     
    end2 = tm.time()
    runtime_py = end2 - start2 
    py_runtime[i] = runtime_py
    
    print("\n",size, ' by ', size)
    print('naive function runtime:', runtime_naive ,'seconds') # printing runtime for naive function
    print('python function runtime:', runtime_py ,'seconds')
    
    
    
```

While the python function stayed at a run time below 1 second, the naive function grew at a very fast and increasing pace as the last iteration neared a minute runtime.

```python

plt.figure()
plt.title("Python function and naive function runtime vs matrix size ")
plt.xlabel('matrix size')
plt.ylabel('time')
plt.plot(matrix_size,naive_runtime,label = "naive runtime")
plt.plot(matrix_size,py_runtime, label = "python func runtime")
plt.legend()
plt.show()
```

My previous observation is shown in this graph as the python function runtime stays nearly constant and near 0 for each iteration. The naive function run time grows larger and larger at an increasing pace.
