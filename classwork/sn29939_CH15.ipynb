```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2 * y

# Define parameters
h_ref = 0.01# step size
x_start_ref, x_end_ref = 0, 2
num_steps_ref = int((x_end_ref - x_start_ref) / h_ref)
x_values_ref = np.linspace(x_start_ref, x_end_ref, num_steps_ref + 1)
y_values_ref = np.zeros(num_steps_ref + 1)
y_values_ref[0]=1

for i in range(num_steps_ref):
    y_values_ref[i + 1] = y_values_ref[i] + h_ref * f(x_values_ref[i], y_values_ref[i])
    

    
h_arr = np.arange(0.01,1.0,0.05)# step size

for h in h_arr:
    x_start, x_end = 0,2
    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)


    # Initial condition
    y_values[0] = 1  # y(0) = 1


    # Implement Euler's method

    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
    
    #errors
    y_ref_interp = np.interp(x_values, x_values_ref, y_values_ref)
    error = np.abs(y_values - y_ref_interp)
    total_error = np.sum(error)
    
    print(f"The error at each point is {error}")
    print(f" The total error for h = {h} is {total_error}")
    
    plt.plot(x_values, y_values, label="Euler's Method Approximation")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title("Approximation of dy/dx = -2y using Euler's Method")
    plt.grid(True)
    plt.show()
    
    
 

# Plot the results
```
