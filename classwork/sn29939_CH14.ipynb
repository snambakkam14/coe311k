```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#actual function

def f(x):
    sin = np.sin(x)
    return sin

#actual derivative

def f_prime(x):
    cos = np.cos(x)
    return cos

a = 0
b = 2 * np.pi
N = 200

x = np.linspace(a,b,N)


# Step size

h = (b - a) / (N - 1)

 

# Central Difference Approximation for the derivative

df_dx_cent = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation

df_dx_forw = (f(x + h) - f(x)) / (h)

# Backward Difference 

df_dx_back = (f(x) - f(x - h)) / (h)


#plotting
plt.title("sin(x)")
plt.plot(x,f(x),color = "blue", label = "sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()



plt.title("different derivative methods")
plt.plot(x, f_prime(x), color = "red" , label=" sin(x) derivative ")
plt.plot(x, df_dx_cent, color = "blue" , label = " central difference approx ")
plt.plot(x, df_dx_forw, color = "green", label = " forward difference approx ")
plt.plot(x, df_dx_back, color = "purple", label = " backward difference approx ")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.legend()
plt.show()


```
